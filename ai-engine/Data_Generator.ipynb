```python
import pandas as pd
from faker import Faker
import random

fake = Faker()

# --- 1. PRODUCT DATASET ---
# Creating a dictionary to store fixed information for each product
product_info = {
    'Tudung Bawal': {'Cost': 15.00, 'Price': 25.00, 'Category': 'Hijab'},
    'Baju Kurung Moden': {'Cost': 45.00, 'Price': 85.00, 'Category': 'Clothing'},
    'Kain Pelikat': {'Cost': 12.00, 'Price': 25.00, 'Category': 'Clothing'},
    'Kurta Slim Fit': {'Cost': 35.00, 'Price': 65.00, 'Category': 'Clothing'},
    'Sampul Raya': {'Cost': 1.50, 'Price': 5.00, 'Category': 'Accessories'},
    'Tudung Shawl': {'Cost': 18.00, 'Price': 35.00, 'Category': 'Hijab'},
    'Baju Melayu': {'Cost': 50.00, 'Price': 120.00, 'Category': 'Clothing'}
}

# --- 2. GENERATE 1000+ ROWS OF SALES DATA ---
num_orders = 1200
data = []
product_list = list(product_info.keys())

for i in range(num_orders):
    product_name = random.choice(product_list)
    qty = random.randint(1, 10)
    
    data.append({
        "Order_Date": fake.date_between(start_date='-1y', end_date='today'),
        "Product_Name": product_name,
        "Category": product_info[product_name]['Category'],
        "Quantity": qty,
        "Cost_Price": product_info[product_name]['Cost'],
        "Unit_Price": product_info[product_name]['Price'],
        "Total_Revenue": qty * product_info[product_name]['Price'],
        "Platform": random.choice(['Shopee', 'TikTok'])
    })

# Convert into DataFrame
df_master = pd.DataFrame(data)
df_master['Order_Date'] = pd.to_datetime(df_master['Order_Date'])
df_master = df_master.sort_values(by="Order_Date")

# --- 3. CALCULATE SALES SUMMARY FOR EACH PRODUCT ---
# Aggregating total units sold and total revenue
summary = df_master.groupby('Product_Name').agg({
    'Quantity': 'sum',
    'Total_Revenue': 'sum'
}).reset_index()

summary = summary.sort_values(by='Quantity', ascending=False)

# --- 4. SAVE RESULTS ---
df_master.to_csv("Sales_History.csv", index=False)

print(f"✅ SUCCESS! File 'Sales_History.csv' with {len(df_master)} rows has been created.")
print("\n--- SALES SUMMARY (TOTAL PER PRODUCT) ---")
print(summary.to_string(index=False))

# Show data sample
print("\n--- SAMPLE OF 5 MASTER DATA ROWS ---")
display(df_master.head())
```
