```python
import pandas as pd
from faker import Faker
import random

fake = Faker()

# --- 1. PRODUCT DATASET ---
# Creating a dictionary to store fixed information for each product
product_info = {
    'CLO-BM-001': {'name': 'Baju Melayu', 'Cost': 50.00, 'Price': 120.00, 'Category': 'Clothing'},
    'CLO-BK-002': {'name': 'Baju Kurung Moden', 'Cost': 45.00, 'Price': 85.00, 'Category': 'Clothing'},
    'CLO-KS-003': {'name': 'Kurta Slim Fit', 'Cost': 35.00, 'Price': 65.00, 'Category': 'Clothing'},
    'HIJ-TB-004': {'name': 'Tudung Bawal', 'Cost': 15.00, 'Price': 25.00, 'Category': 'Hijab'},
    'HIJ-TS-005': {'name': 'Tudung Shawl', 'Cost': 18.00, 'Price': 35.00, 'Category': 'Hijab'},
    'CLO-KP-006': {'name': 'Kain Pelikat', 'Cost': 12.00, 'Price': 25.00, 'Category': 'Clothing'},
    'ACC-SR-007': {'name': 'Sampul Raya', 'Cost': 1.50, 'Price': 5.00, 'Category': 'Accessories'}
}

# --- 2. GENERATE 1000+ ROWS OF SALES DATA ---
num_orders = 1200
data = []
product_list = list(product_info.keys()) # This now contains a list of SKUs

for i in range(num_orders):
    sku_code = random.choice(product_list) # Randomly selecting SKUs
    qty = random.randint(1, 10)
    
  # Follow the new column name 
    data.append({
        "order_date": fake.date_between(start_date='-1y', end_date='today'),
        "sku": sku_code,
        "quantity_sold": qty,
        "platform": random.choice(['Shopee', 'TikTok'])
    })

# Convert into DataFrame
df_master = pd.DataFrame(data)
df_master['order_date'] = pd.to_datetime(df_master['order_date'])
df_master = df_master.sort_values(by="order_date")


df_master.to_csv('Sales_History.csv', index=False)
print("Sales_History.csv SKU version successfully exported!")

# --- 3. CALCULATE SALES SUMMARY FOR EACH PRODUCT ---
# Aggregating total units sold and total revenue
#summary = df_master.groupby('Product_Name').agg({
  #  'Quantity': 'sum',
 #   'Total_Revenue': 'sum'
#}).reset_index()

#summary = summary.sort_values(by='Quantity', ascending=False)

# --- 4. SAVE RESULTS ---
#df_master.to_csv("Sales_History.csv", index=False)

#print(f"✅ SUCCESS! File 'Sales_History.csv' with {len(df_master)} rows has been created.")
#print("\n--- SALES SUMMARY (TOTAL PER PRODUCT) ---")
#print(summary.to_string(index=False))

# Show data sample
#print("\n--- SAMPLE OF 5 MASTER DATA ROWS ---")
#display(df_master.head())
```
